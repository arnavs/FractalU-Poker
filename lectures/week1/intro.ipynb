## Writing a Fixed Point Solver

```julia
f(x) = 0.1 + 0.9 * x
```

```julia
x0 = 0.5
```

```julia
x_old = x0 
for i in 1:100
    x_new = f(x_old)
    @show x_new 
    x_old = x_new 
end
```

```julia
function fixedpointmap(f, iv, tolerance, maxiter)
    # setup the algorithm
    x_old = iv
    normdiff = Inf
    iter = 1
    while normdiff > tolerance && iter <= maxiter
        x_new = f(x_old) # use the passed in map
        normdiff = abs(x_new - x_old)
        x_old = x_new
        iter = iter + 1
    end
    return (x_old, normdiff, iter)
end
```

```julia
fixedpointmap(x -> 0.2 + 2 * x, 0.5, 1e-10, 300)
```

```julia
g = x -> 0.2 + 2 * x
x_old = x0 
for i in 1:100
    x_new = g(x_old)
    @show x_new 
    x_old = x_new 
end
```
